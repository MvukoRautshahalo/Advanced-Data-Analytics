# Data Definition Language

## Create schema / Database

```sql
CREATE DATABASE IF NOT EXISTS BRIGHTCOFFEE_20250519_V2
COMMENT 'Bright Coffee Shop Analytics - Database';
```

Database.schema.table

dim_customers 
- customer_id (PK) 
- first_name 
- last_name 
- email 
- tier 
- loyalty_points 

```sql
CREATE TABLE IF NOT EXISTS BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
(
    customer_id BIGINT PRIMARY KEY,
    FIRST_NAME VARCHAR(50),
    LAST_NAME STRING,
    email STRING,
    tier string,
    loyalty_points int
)
```

FACT TABLE 
fact_orders 
- order_id (PK) 
- customer_id (FK) 
- product_id (FK) 
- store_id (FK) 
- date_id (FK) 
- quantity 
- unit_price 
- total_amount 

```sql
CREATE TABLE IF NOT EXISTS BRIGHTCOFFEE_20250519_V1.FACT_ORDERS
(
    order_id BIGINT PRIMARY KEY, 
    customer_id BIGINT, 
    product_id BIGINT,
    store_id BIGINT,
    ORDER_date DATE,
    quantity INT,
    unit_price DECIMAL(8,2), 
    total_amount DECIMAL (10,2)
)
USING DELTA
PARTITIONED BY (ORDER_DATE)
```

## DROP TABLE

```sql
DROP TABLE BRIGHTCOFFEE_20250519_V1.FACT_ORDERS;
```

## create backup

```sql
create table BRIGHTCOFFEE_20250519_V1.FACT_ORDERS_backup 
as select * from BRIGHTCOFFEE_20250519_V1.FACT_ORDERS;
```

## ALTER TABLE

```sql
ALTER TABLE BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS ADD COLUMN PHONE STRING
```

```sql
--Allows you to rename columns
ALTER TABLE BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS SET TBLPROPERTIES('delta.columnMapping.mode'='name')
```

```sql
ALTER TABLE BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS RENAME COLUMN PHONE TO PHONE_NUMBER
```

# Data Manipulation Language

## insert data

```sql
insert into BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS (customer_id, FIRST_NAME, LAST_NAME, email, tier)
values (1, "John", "Smith", "john@gmail","Gold")
```

```sql
select * from BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
```

## insert multiple values

```sql
insert into BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS (customer_id, FIRST_NAME, LAST_NAME, email, tier)
values (2, "Thandi", "Mokoena", "thandi@gmail","Gold"),
(3, "Siya", "Dlamini", "siya@gmail","Gold")
```

```sql
select * from BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
```

## update data

```sql
update BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
set tier = 'Silver', --Value changing
loyalty_points = 50
where FIRST_NAME = "John"
```

```sql
select * from BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
```

## Delete data

```sql
delete from BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
where customer_id = 2
```

```sql
select * from BRIGHTCOFFEE_20250519_V1.DIM_CUSTOMERS
```
