# Data Definition Language & Data Manipulation Language

# Fact & dimension table - Star Schema

| Concept              | Description                                                                                   | Coffee Shop Example                                              |
| -------------------- | --------------------------------------------------------------------------------------------- | ---------------------------------------------------------------- |
| **Primary Key (PK)** | Uniquely identifies each row. No two rows can share the same PK.                              | `dim_customers.customer_id`                                      |
| **Foreign Key (FK)** | A column in one table that refers to the PK of another table, creating a relationship.        | `fact_orders.customer_id` references `dim_customers.customer_id` |
| **Measure**          | A numeric value in the fact table that can be aggregated (summed, averaged, counted).         | `fact_orders.total_amount`, `quantity`                           |
| **Attribute**        | A descriptive column in a dimension table used for filtering, grouping, or categorizing data. | `dim_products.category`, `dim_customers.tier`                    |


# Data Definition Language

DDL Commands
|Command| What It Does| Reversible?|
|-|-|-|
|CREATE SCHEMA| Creates a new schema/database container |No|
|CREATE TABLE |Creates a new table with defined columns |No|
|CREATE TABLE AS SELECT |Creates and fills a table from a query |No|
|ALTER TABLE ADD COLUMN| Adds a new column to an existing table |Drop the column|
|ALTER TABLE DROP COLUMN| Removes a column permanently |Data is lost|
|ALTER TABLE RENAME COLUMN| Renames an existing column |Rename back|
|DROP TABLE |Removes the table and all data |Delta time travel only|

# Data Munipulation Language

DML Commands
|Command |What It Does|
|-|-|
|INSERT INTO ... VALUES| Adds specific rows to a table|
|INSERT INTO ... SELECT| Copies rows from a query into a table|
|UPDATE ... SET ... WHERE| Modifies existing rows matching a condition|
|DELETE FROM ... WHERE| Removes rows matching a condition|

# Constraints & Data integrity

| Constraint      | What It Enforces                                           | Example                              |
| --------------- | ---------------------------------------------------------- | ------------------------------------ |
| **NOT NULL**    | Column must always have a value — no blanks allowed.       | `first_name STRING NOT NULL`         |
| **UNIQUE**      | No two rows can share the same value in this column.       | `CONSTRAINT uq_email UNIQUE (email)` |
| **CHECK**       | Value must pass a custom condition you define.             | `CHECK (price > 0)`                  |
| **DEFAULT**     | Fallback value used when no value is provided on `INSERT`. | `tier STRING DEFAULT 'Bronze'`       |
| **PRIMARY KEY** | Unique and `NOT NULL` combined — identifies each row.      | `customer_id BIGINT PRIMARY KEY`     |
