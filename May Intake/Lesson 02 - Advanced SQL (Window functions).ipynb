A window function performs a calculation across related rows without collapsing them or losing details

function() over (partition by column order by column) as calc


sum(sales) over (partition by transaction_date order by date) as running_total

```sql
%sql

select 
transaction_date,
sum(transaction_qty*unit_price) sales,

sum(sum(transaction_qty*unit_price)) over (order by transaction_date asc) as running_sales

from brightlearn.brightcoffee.shop_sales
group by transaction_date
```

## How is each store performing over time?
- partition to split data into min datasets
- Groups rows in a dataset based on certain column


Store_location, date, sales

partiton by store

mini datasets per store location




```sql
%sql
select 
store_location,
transaction_date,

sum(transaction_qty*unit_price) daily_sales,

sum(sum(transaction_qty*unit_price)) over (
    partition by store_location 
    order by transaction_date asc
    ) as running_sales_per_store
    
from brightlearn.brightcoffee.shop_sales
group by store_location, transaction_date;
```

WHta are our best performing products

```sql
%sql

select 
product_type,

sum(transaction_qty*unit_price) sales,

rank() over (order by sum(transaction_qty*unit_price) desc
) as product_rank

from brightlearn.brightcoffee.shop_sales
where 
group by product_type
```
