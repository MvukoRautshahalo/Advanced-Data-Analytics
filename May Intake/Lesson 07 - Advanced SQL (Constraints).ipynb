# Constraints
|Constraint |What It Enforces |Example
|-|-|-|
|NOT NULL| Column must always have a value - no blanks allowed|first_name STRING NOT NULL|
|UNIQUE|No two rows can share the same value in this column|CONSTRAINT uq_email UNIQUE (email)|
|CHECK |Value must pass a custom condition you define|CHECK (price > 0)|
|DEFAULT|Fallback value used when no value is provided on INSERT|tier STRING DEFAULT |'Bronze'|
|PRIMARY KEY| Unique and NOT NULL combined - identifies each row|customer_id BIGINT PRIMARY KEY|
